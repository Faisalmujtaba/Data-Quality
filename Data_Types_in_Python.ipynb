**3. Creating list of mixed data types and sorting them.**

```python
mixed_list = ['Orange', 10, 4.5, 'Banana', 2, 'Grapes']

def custom_key(item):
    if isinstance(item, int):
        return 0, item
    elif isinstance(item, str):
        return 1, item
    elif isinstance(item, float):
        return 2, item

sorted_list = sorted(mixed_list, key=custom_key)

print(sorted_list)
```

**Data Types**

**1. Creating Variable of different data types and performing operations.**

```python
integer_var = 10
float_var1 = 5.5
float_var2 = 3.7
string_var1 = "Hello"
string_var2 = "World"
boolean_var = True

# Perform operations
addition = integer_var + float_var1
subtraction = float_var2 - integer_var
concatenation = string_var1 + " " + string_var2
logical = boolean_var and True

# Print results
print("Addition result:", addition)
print("Subtraction result:", subtraction)
print("Concatenation result:", concatenation)
print("Logical result:", logical)
```

**2. Converting a string to uppercase or lowercase**
